```python
def split_csv(input_file, chunk_size=200, columns = None):
    main_file = pd.read_csv(input_file)

    if columns:
        main_file = main_file[columns]

    total_chunks = (len(main_file) // chunk_size) + (1 if len(main_file) % chunk_size !=0 else 0) 

    print(f"Total rows: {len(main_file)}")
    print(f"Total chunks: {total_chunks}")

    for i in range(3):
        chunk = main_file[i * chunk_size : (i+1) * chunk_size]  #slicing the main file.
    
        output_file = f"../raw_data/sales_data_{i}.csv"
    
        chunk.to_csv(output_file, index=False)
        print(f"Exported sucessfully: {output_file}")

col_to_select = ["Product", "Amount","Boxes"]
split_csv("combined_file.csv", chunk_size = 200, columns= col_to_select)
    
```

```python
def combine_chunks(folder_path):
    all_files = [f for f in os.listdir(folder_path) if f.endswith('.csv')]

    data_file = pd.DataFrame()

    for file_name in all_files:
        df = pd.read_csv(os.path.join(folder_path, file_name))
        data_file = pd.concat([data_file, df], ignore_index = True)

    return data_file
```

```python
def log_capture(file_names):
    with open("../exports/log.txt", "a") as log_file:
        log_file.write(f"{file_names}\n")
```

```python
#this function is used for excel(.xls) files exporting.

def process_sales_batch(folder_path, export_folder="../exports"):

    df = combine_chunks(folder_path)
    grup  = df.groupby("Product")[["Amount","Boxes"]].sum().reset_index()
    #group = grup[grup['Amount']>30000]
    
    print("Daily Sales Summary\n" + '_'*60)
    for _, row in grup.iterrows():
        print(f"Product: {row['Product']}, Amount: {row['Amount']}, Boxes: {row['Boxes']}")
        
    #export

    os.makedirs(export_folder, exist_ok=True)
    date_str = datetime.today().strftime("%Y-%m-%d")
    filename = f"{export_folder}/Sales_Summary_Batch_{date_str}.xlsx"
    log_capture(filename)
    grup.to_excel(filename, index=False)
    print(f"\n Exported summary to: {filename}")
```
